# Solar Filament Inference Notebook
This notebook loads a trained UNet model and generates predictions for the solar filament segmentation task.

```python
# Import required libraries for inference
import os
import sys
import yaml
import numpy as np
import jax
import jax.numpy as jnp
import equinox as eqx

# Ensure repo is cloned
if not os.path.exists("refactored-potato"):
    os.system("git clone https://github.com/12in-the-night/refactored-potato.git")
sys.path.append("refactored-potato/src")

# Import model and generate function
from main import UNet, generate_submission

# Override config for Kaggle environment
import main
main.IS_KAGGLE = True
main.BASE = "/kaggle/input/filament-segmentation-2026/MAGFiLO_1.0_Kaggle_2026"
main.cfg['paths']['test_img'] = "test/test_images"

# Load weights
key = jax.random.PRNGKey(0)
model = UNet(in_channels=1, out_channels=1, base=main.BASE_CH, key=key)
model = eqx.tree_deserialise_leaves("refactored-potato/ema_best.eqx", model)

# Generate submission
generate_submission(model)
```
